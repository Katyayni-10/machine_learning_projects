```python
import gymnasium as gym
import numpy as np
import pickle
import time
```

```python
with open("q_table.pkl", "rb") as file:
    q_table = pickle.load(file)
print("Q-Table loaded successfully")
print("Shape :", q_table.shape)
```

```python
env = gym.make("Taxi-v4", render_mode="ansi")
```

```python
state, info = env.reset()
total_reward = 0
done = False
step = 1
while not done:
    print("=" * 50)
    print(f"Step : {step}")
    print(env.render())
    action = np.argmax(q_table[state])
    action_names = {
        0: "Move South",
        1: "Move North",
        2: "Move East",
        3: "Move West",
        4: "Pickup Passenger",
        5: "Drop Passenger"
    }
    print("Action :", action_names[action])
    next_state, reward, terminated, truncated, info = env.step(action)
    done = terminated or truncated
    print("Reward :", reward)
    total_reward += reward
    state = next_state
    step += 1
    time.sleep(0.5)
print("=" * 50)
print("Episode Finished!")
print("Total Reward :", total_reward)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10,5))
plt.plot(total_rewards, marker='o')
plt.title("Testing Rewards")
plt.xlabel("Episode")
plt.ylabel("Reward")
plt.grid(True)
plt.show()
```
